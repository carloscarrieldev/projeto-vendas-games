# Distribuição

```python
plt.figure(figsize = (10, 6))
plt.hist(df['Global_Sales'], bins=30, color='Black')
plt.title('Distribuição de Vendas')
plt.xlabel('Vendas')
plt.ylabel('Quantidade')
plt.show()
```

# Resultado
- A análise mostra que muitas vendas estão concentradas em produtos com maiores quantidades, enquanto outros apresentam resultados muito abaixo do esperado. 

# Dispersão

```python
plt.figure(figsize = (10,6))
plt.scatter(y = df['NA_Sales'], x = df['EU_Sales'], color='Green')
plt.title('Dispersão entre America x Norte')
plt.ylabel('Vendas - America')
plt.xlabel('Vendas - Europa')
plt.show()
```

# Resultado
- Após a análise, o gráfico mostra que as duas regiões estão bem equilibradas, apresentando o mesmo desempenho, mas a América do norte mostrou ter vendido uma quantidade maior.

# Outliers

```python
plt.figure(figsize= (10, 6))
plt.boxplot(df['Global_Sales'])
plt.title('Outliers nas Vendas')
plt.ylabel('Vendas')
plt.ylim(0, 5)
plt.show()
```

# Resultado
- Após análise, mostra resultados de vendas acima da média, com outliers elevados indicando que os jogos tiveram ótimo desempenho.
